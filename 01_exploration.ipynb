# Steel Industry Energy Consumption — ML Regression Benchmark

Projet collaboratif (7 personnes) autour du dataset UCI **Steel Industry Energy Consumption**.  
Objectif : comparer plusieurs algorithmes de régression pour **prédire la consommation d’énergie (Usage_kWh)** à partir de variables process + temps + catégories.

## Dataset
- Source : UCI Machine Learning Repository — *Steel Industry Energy Consumption*
- Volume : ~35k lignes (données horodatées)
- Cible (target) : `Usage_kWh`

## Méthodologie
- Nettoyage + préparation des données
- Feature engineering temps à partir de `date` :
  - `hour`, `minute`, `day`, `month`, `dayofweek`
- Encodage des variables catégorielles :
  - `WeekStatus`, `Day_of_week`, `Load_Type` (One-Hot)
- Split **chronologique** (éviter data leakage) :
  - train = 80% / test = 20% (les dernières lignes = test)
- Validation :
  - `TimeSeriesSplit` (cross-validation sur train)
- Métriques :
  - **MAE**, **RMSE**, **R²**

## Modèles testés
- KNN → `KNeighborsRegressor`
- SVM → `SVR`
- Arbre → `DecisionTreeRegressor`
- Linéaire → `LinearRegression`
- Polynôme → `PolynomialFeatures` + `LinearRegression`
- Ensemble 1 → `RandomForestRegressor`
- Ensemble 2 → `GradientBoostingRegressor`
- (Optionnel) XGBoost → `XGBRegressor`

## Livrables
- Tableau comparatif des modèles (MAE / RMSE / R²)
- Modèles sauvegardés (`joblib`)
- Rapport reproductible dans `reports/`
